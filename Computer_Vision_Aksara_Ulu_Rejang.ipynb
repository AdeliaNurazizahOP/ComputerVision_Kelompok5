Training model AI (YOLO12) agar bisa mengenali dan mendeteksi Aksara Ulu Rejang

```python
from ultralytics import YOLO

# Muat model dasar YOLO12
model = YOLO('yolo12s.pt')

# Mulai training
# dataset.location otomatis mendeteksi lokasi file data.yaml dari Roboflow
results = model.train(
    data=f"{dataset.location}/data.yaml",
    epochs=50,      # Jumlah iterasi pembelajaran karakter aksara
    imgsz=640,      # Resolusi gambar
    batch=16,
    name='model_aksara_kaganga'
)
```
